```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
image = cv2.imread("content.jpg")
cv2_imshow(image)
```

```python
print("SHAPE:", image.shape)
# 636 => height
# 994 => width
# 3   => color channels (BGR) => Blue, Green, Red
print("HEIGHT:", image.shape[0])
print("WIDTH:", image.shape[1])
print("CHANNELS:", image.shape[2])
```

```python
(b, g, r) = image[100, 50]
print("Blue:", b, "Green:", g, "Red:", r)
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray)
```

```python
resized = cv2.resize(gray,(300, 300))
cv2_imshow(resized)
```

```python
cropped = image[100:300, 150:350]
cv2_imshow(cropped)
```

```python
rotated = cv2.rotate(gray, cv2.ROTATE_180)
cv2_imshow(rotated)
```

```python
flipped = cv2.flip(gray, 1)
cv2_imshow(flipped) #flip bu rasmni teskariga ogiradi
```

```python
cv2.imwrite("flippedimage.jpg", flipped)
```
