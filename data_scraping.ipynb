```python
import os, shutil

source_dir = '/content/pilaf_images/uzbek pilaf plov food' # ayni vaqtdagi rasmlar joylashgan file nomi
target_dir = '/content/drive/MyDrive/food101_dataset/pilaf' # qayerga borib joylashsin

os.makedirs(target_dir, exist_ok=True) #googledrivega borib targetdir ga mos yangi folder yaratib googledan yuklab olgan rasmlarimizni osha yerga joylaymiz.exist_ok bu agar shu nomli folder avvalda mavjud bolsa error berma degan mano

for filename in os.listdir(source_dir):
    src_file = os.path.join(source_dir, filename) #
    dst_file = os.path.join(target_dir, filename)
    shutil.move(src_file, dst_file)
```
